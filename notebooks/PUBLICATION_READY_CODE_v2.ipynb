```python
%cd /content/drive/MyDrive/PUBLICATION_READY_CODE/
from data.build import build_dataloaders
from models.model import DINOProtoFormer
from configs.config import *

_, _, test_loader = build_dataloaders(
    data_root=DATA_ROOT,
    batch_size=BATCH_SIZE,
    num_workers=NUM_WORKERS,
    image_size=IMAGE_SIZE
)

model = DINOProtoFormer().to(DEVICE)

checkpoint = torch.load(
    os.path.join(CHECKPOINT_DIR, "best_model.pth"),
    map_location=DEVICE
)

model.load_state_dict(checkpoint["model"])

model.eval()

images, masks = next(iter(test_loader))

images = images.to(DEVICE)
masks = masks.to(DEVICE)

with torch.no_grad():
    outputs = torch.sigmoid(model(images))

print("Output Statistics")
print("-----------------")
print("Output min :", outputs.min().item())
print("Output max :", outputs.max().item())
print("Output mean:", outputs.mean().item())

preds = (outputs > 0.5).float()

print("\nPixel Statistics")
print("-----------------")
print("Ground Truth Pixels :", masks.sum().item())
print("Predicted Pixels    :", preds.sum().item())
```
